Extra 1: Control de aforo con break y continue (while)
Crea un programa que procese visitantes del museo con un ciclo while. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa continue). Si el total acumulado alcanza o supera $100, detén el ciclo con break y muestra el total. Entrada:

Costo del boleto 1: 45
Costo del boleto 2: 0
Costo del boleto 3: 30
Costo del boleto 4: 45

```python
total = 0
boletos = 1
while True:
    precio = int(input(f"Costo de boleto {boletos}: "))
    boletos += 1

    if precio == 0:
        print("MENOR DE 3 AÑOS, SIN CARGO")
        continue
    total += precio

    if total >= 100:
        print("AFORO ALCANZADO, SE DETIENE EL REGISTRO")
        break

print(f"TOTAL ACOMULADO {total:.2f}")

```

Extra 2: Estadística de visitantes con for
Crea un programa que pida el número total de visitantes y, con un ciclo for, capture la edad de cada uno. Debe contar cuántos son adultos (18 años o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos resultados con dos decimales. Entrada:

Número de visitantes: 4
Edad del visitante 1: 30
Edad del visitante 2: 45
Edad del visitante 3: 15
Edad del visitante 4: 2

Salida:

Adultos: 2
Promedio de edad: 23.00

```python
mayor_edad=0
suma_edades=0
visitantes = int(input("INGRESE LA CANTIDAD DE VISITANTES: "))
for i in range(1,visitantes +1):
    edad=int(input(f"Ingrese la edad del visitante: {i} "))
    suma_edades += edad
    if (edad >= 18):
        mayor_edad +=1
promedio = suma_edades / visitantes
print(f"\nAdultos: {mayor_edad}")
print(f"El promedio de los mayores de edad {promedio:.2f}")
```

Extra 4: Pirámide de asteriscos con for
Crea un programa que pida la altura de una pirámide y la dibuje con asteriscos usando ciclos for anidados y el parámetro end de print. Entrada:

Altura de la pirámide: 4

Salida:

*
**
***
****

```python
numero=int(input("Ingrese el numero de altura para su piramide: "))

for i in range(1, numero +1):
    print("*" * i)
```
