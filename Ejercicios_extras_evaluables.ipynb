Extra 1: Control de aforo con break y continue (while)
En este programa utilize while True para en ingreso de precios e incrementar hasta el numero deseado y el break para interumpir el ciclo

```python
total = 0
boletos = 1
while True:
    precio = int(input(f"Costo de boleto {boletos}: "))
    boletos += 1

    if precio == 0:
        print("MENOR DE 3 AÑOS, SIN CARGO")
        continue
    total += precio

    if total >= 100:
        print("AFORO ALCANZADO, SE DETIENE EL REGISTRO")
        break

print(f"TOTAL ACOMULADO {total:.2f}")

```

Extra 2: Estadística de visitantes con for
En este programa utilizamos el for para pedir la edad de cada visitante y acomularlas en una suma de esdades y ver la cantidad de adultos.

```python
mayor_edad = 0
suma_edades = 0
visitantes = int(input("INGRESE LA CANTIDAD DE VISITANTES: "))
for i in range(1,visitantes +1):
    edad = int(input(f"Ingrese la edad del visitante: {i} "))
    suma_edades += edad
    if (edad >= 18):
        mayor_edad += 1
promedio = suma_edades / visitantes
print(f"\nAdultos: {mayor_edad}")
print(f"El promedio de los mayores de edad {promedio:.2f}")
```

Extra 4: Pirámide de asteriscos con for
en este programa se pide un numero al usuario que define la altura de la piramide y con el for va sumando de 1 en cada nivel diferente hasta el numero ingresado

```python
numero=int(input("Ingrese el numero de altura para su piramide: "))

for i in range(1, numero +1):
    print("*" * i)
```
