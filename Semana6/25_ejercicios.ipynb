Ejercicio 10: Suma acumulada con while
Pide números mientras sean distintos de 0 (el 0 termina el ciclo y no se suma). Entrada: 5, 8, 2, 0 → Salida: La suma de los números es: 15

Ejercicio 14: Máximo de una lista con función que regresa valor (sin max())
Entrada: [7, 3, 9, 2] 

```python
def encontrar_maximo(lista):
    mayor = lista[0]
    for numero in lista:
        if numero > mayor:
            mayor = numero
    return mayor

numeros = [7, 13, 9, 2]
resultado = encontrar_maximo(numeros)

print(f"El maximo de la lista es: {resultado}")
```

Ejercicio 15: Matriz y suma de la diagonal principal
Entrada: [[1,2,3],[4,5,6],[7,8,9]] → Salida: La suma de la diagonal principal es: 15

```python
matriz = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
suma = 0
for fila in range(len(matriz)):
    suma += matriz[fila][fila]

print(f"La suma principal es: {suma}")
```

Ejercicio 18: Factorial con función recursiva
Entrada: 5 → Salida: El factorial de 5 es: 120

```python
def factorial(numero): 
    if numero == 0 or numero == 1: 
        return 1 
    return numero * factorial(numero - 1) 
numero = int(input("Ingrese un número: ")) 
resultado = factorial(numero) 
print(f"El factorial de {numero} es: {resultado}")
```

Ejercicio 21: Diccionario de calificaciones y promedios
Entrada: {"Ana": [8, 9, 10], "Beto": [7, 7, 8]}` → Salida:
Ana: 9.00
Beto: 7.33

```python
calificaciones = { "Ana": [8, 9, 10], "Beto": [7, 7, 8] } 
for nombre, notas in calificaciones.items(): 
    suma = 0 
    for nota in notas: 
        suma += nota 
    promedio = suma / len(notas) 
    print(f"{nombre}: {promedio:.2f}")
```
