Ejercicio 1: Mayor de edad
Enunciado: Crea un programa que pida la edad de una persona y muestre si es mayor de edad (18 años o más) o menor de edad. 
Entrada: 17 Salida: Eres menor de edad

```python
edad=int(input("Ingrese su edad"))

if edad >= 18 :
    print("Eres mayor de edad")
else:
    print("Aun eres menor de edad")
```

Enunciado: Crea un programa que pida un número entero y determine si es par o impar usando el operador módulo %. Entrada: 7 Salida: 7 es un número impar

```python
numero=int(input("Ingresa el numero"))

if (numero % 2):
    print("El numero es impar")
else:
    print("El numero es par")
```
