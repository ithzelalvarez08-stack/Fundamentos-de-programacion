Ejercicio 5:
Crea un programa que pida el tipo de visitante (adulto_mayor, profesor, estudiante u otro) y calcule el precio de un boleto base de $45 aplicando solo un descuento: adulto mayor 12%, profesor 10%, estudiante 10%, otro 0%. 
Entrada: estudiante Salida: Precio final: $40.50

```python
visitante=input("Ingresa el tipo de visitante que eres")

precio_normal=45
if (visitante =="adulto_mayor"):
    descuento= 0.12
elif (visitante=="profesor" or visitante=="estudiante"):
    descuento=0.10
else:
    descuento=0
monto_descuento= (precio_normal*descuento)
precio_final= precio_normal - monto_descuento
print(f"El precio total es: ${precio_final:.2f}")
```
