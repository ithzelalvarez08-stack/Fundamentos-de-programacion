Extra 1: División de cuenta con propina.
Despues de analizar el problema, asigne las variables como:
"tdcuenta" para saber cual es el pago de cuenta inicial.
"porcentaje" nos indica cual es el porcentaje empleado en la propina.
"tcp" es para saber el total de la cuenta ya con la propina.
En este caso utilize el comando "float" ya que el resultado nos puede dar en decimal.


```python
tdcuenta=float(input("Escriba el total de la cuenta: "))
porcentaje=float(input("Escriba el porcentaje de propina: "))
personas=float(input("Escriba cuantas personas son: "))

propina= (tdcuenta * (porcentaje / 100))
tcp= (tdcuenta + propina)
pagoporpersona= (tcp / personas)

print("La propina es $",propina)
print("Total con propina: $",tcp)
print("Pago por persona: $",pagoporpersona)
```

 Extra 2: Conversor de minutos a días, horas y minutos.
 Despues de comprender el problema, asigne las variables para el desarrollo del programa como lo son:
 "tminutos" son los minutos que el usuario va ingresar para convertir.
 Se utilizaron comandos como "float" pero tembien en las operaciones se utilizo "//" que nos indica que la divison sea entera.

```python
tminutos=float(input("Escriba los minutos: "))
dias= (tminutos // 1440)
mrestantes= (tminutos % 1440)
horas= (mrestantes // 60)
minutos= (mrestantes % 60)
print(tminutos,"= ",dias,"dia(s), ",horas,"hora(s)",minutos,"minutos(s)")

```

Extra 3: Calificación final ponderada.
Al igual que los programas utilize el comando "float" y al momento de calcular el promedio por parcial se cambio el porcentaje que valia cada parcial.

```python
parcial1=float(input("Escriba su calificacion de Parcial 1: "))
parcial2=float(input("Escriba su calificacion de Parcial 2: "))
parcial3=float(input("Escriba su calificacion de Parcial 3: "))

calificacionfinal= ((parcial1 * 0.30) + (parcial2 * 0.30) + (parcial3 * 0.40))

print ("Tu calificacion final es: ", calificacionfinal)
```
