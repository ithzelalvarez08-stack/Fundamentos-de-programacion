Extra 1: Suma de todos los elementos de una matriz
Crea una función suma_matriz(matriz) que regrese valor (con return) la suma de todos los elementos de una lista de listas, usando ciclos for anidados. Define además una función imprimir_matriz(matriz) que no regrese valor y muestre la matriz sin corchetes ni comas. Aplica ambas funciones a la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9]]. Entrada:

```python
matriz = [
    [3,4,5],
    [6,7,8],
    [1,2,9]
]
suma = 0
for f in range(len(matriz)):
    for c in range(len(matriz[f])):
        print(matriz[f][c], end = " ")
        suma = suma + matriz[f][c]
    print()
print("La suma de la matriz es: ", suma)
```

Extra 3: Producto de una columna sin usar el operador *
Crea la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]] (4 filas y 3 columnas). Define una función multiplicar_columna(matriz, columna) que regrese valor (con return) el producto de los elementos de una columna dada. Restricción: dentro de la función está prohibido usar el operador *; el producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el número de columna (0, 1 o 2) y muestra el resultado.

```python
matriz= [
    [1,2,3],
    [4,5,6],
    [7,8,9],
    [2,4,6]
]
def multiplicar_columna(matriz,columna):
    elementos = [f[columna] for f in matriz]
    resultado = elementos [0]
    for n in elementos[1:]:
        acumulado = 0
        for _ in range(n):
            acumulado += resultado
        resultado = acumulado
    return resultado

columna = int(input("Ingresa el número de columna: "))
print(f"El producto de la columna {columna} es: {multiplicar_columna(matriz, columna)}")
```

Extra 4: Tabla de Pitágoras de tamaño variable
Crea una función generar_tabla(tamano) que regrese valor (con return) una lista de listas de tamano x tamano con los productos de la tabla de Pitágoras (renglones y columnas del 1 al tamano). Dentro de la función de generación está prohibido usar el operador *; usa sumas repetidas con ciclos. Define una función imprimir_tabla(tabla) que no regrese valor e imprima la tabla sin corchetes ni comas, con los números tabulados (\t). Pide al usuario un tamaño entre 2 y 5; si ingresa un valor fuera del rango, muestra un mensaje de error. Al final pide un renglón y una columna y muestra el producto consultando la matriz, sin usar *. Entrada:

```python
def generar_tabla(tamaño):
    tabla = []
    for fila in range(1, tamaño + 1):
        renglon = []
        for columna in range(1, tamaño + 1):
            resultado = 0
            for i in range(columna):
                resultado = resultado + fila
            renglon.append(resultado)
        tabla.append(renglon)
    return tabla

def imprimir_tabla(tabla):
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t")
        print()

tamaño = int(input("Ingrese el tamaño de la tabla: "))

if tamaño < 2 or tamaño > 5:
    print("Error: el tamaño debe estar entre 2 y 5.")
else:
    tabla = generar_tabla(tamaño)
    imprimir_tabla(tabla)
    
    renglon = int(input("Renglon (factor): "))
    columna = int(input("Columna (factor): "))
    
    resultado = tabla[renglon - 1][columna - 1]
    print(f"El producto de {renglon} x {columna} es:  {resultado}")
```
