```python
def suma_tupla(t):
    total = 0
    for e in t:
        total += e
    return total

calorias = (800, 950, 700, 1300, 800, 900, 650)

resultado = suma_tupla(calorias)

print(f"Metiste a tu cuerpo: {resultado} calorias")
```

```python
contactos = {
    "Juan": "555-0101",
    "Luis": "555-0102",
    "Ale": "555-0103"
}

contactos ["Pedro"] = "63888389"

contactos["Ale"] = "7829902"

print(contactos)
```

```python
print(contactos.keys())
print(contactos.values())
print(contactos.items())
```

```python
print("\nimpresion de llaves")
for n in contactos.keys():
    print(n)

print("\nimpresion de valores")
for v in contactos.values():
    print(v)

print("\nimpresion de completa")
for n, t in contactos.items():
    print(f"Nombre: {n}, telefono {t}")
```

```python
def buscar_contacto ( contactos, nombre):
    return contactos.get(nombre, "Contacto no encontrado")
nombre = input("Escribe el nombre del contacto")
telefono = buscar_contacto(contactos, nombre)
print(f"El telefono de {nombre} es: {telefono} ")
```

Excepciones

```python
try:
    edad = int(input("Ingresa tu edad: "))
    print(f"Edad registrada: {edad}")
except ValueError:
    print("Error: Debes ingresar solo números enteros.")
```

```python
try:
    numero = int(input("Ingresa un número entero: "))
    print("El doble es:", numero * 2)
except ValueError:
    print(" El dato ingresado no es un número entero.")
```

```python
try:
    a = int(input("Ingresa el dividendo: "))
    b = int(input("Ingresa el divisor: "))
    print("El resultado de la división es:", a / b)
except ValueError:
    print("Error: Debes ingresar números enteros.")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero.")
```
