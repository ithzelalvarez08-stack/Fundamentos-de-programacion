Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.

```python
candidatos = {
    "A":0,
    "B":0,
    "C":0
}
cantidad = int(input("Cuantos votos van a ser: "))
conteo = 1


while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A,B o C): ")
    if voto.upper() in candidatos:
         print(f"Voto valido para {voto.upper()}")
         candidatos[voto.upper()] +=1
         conteo +=1
    else:
        print("El voto no es valido, vota nuevamente")
print("\nResultados: ")
for k, v in candidatos.items():
    print(f"Opcion {k}, votos: {v}")
```

Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque try-except. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo while. Luego pide la división del primero entre el segundo, con un bloque específico para ZeroDivisionError.

```python
while True:
    try:
        a = int(input("Ingresa el primer numero: "))
        b = int(input("Ingresa el segundo numero: "))
        division= a / b
    except ValueError:
        print("Error: Debes ingresar números enteros.")
    except ZeroDivisionError:
        print("Error: No es posible dividir entre cero.")
    else:
        print(f"El resultado de la division es: {division} ")
        break
```

Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con len(), la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
def contar_palabras(frase):
    palabras = frase.split()
    return len(palabras)


frase = input("Ingrese una frase de su preferencia: ")

print("Longitud de la frase:", len(frase))

print("Frase en mayusculas:")
print(frase.upper())

nueva_frase = frase.replace("Python", "programación")

print("Frase modificada:")
print(nueva_frase)

cantidad_palabras = contar_palabras(frase)

print("Cantidad de palabras:", cantidad_palabras)
```

Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo while y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario contactos, (3) dividir dos números con manejo de ZeroDivisionError y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
def contar_palabras(frase):
    palabras = frase.split()
    return len(palabras)


def analizar_texto():
    frase = input("Ingrese una frase: ")

    print("\nLongitud de la frase:", len(frase))

    print("Frase en mayúsculas:")
    print(frase.upper())

    nueva_frase = frase.replace("Python", "programación")

    print("Frase modificada:")
    print(nueva_frase)

    cantidad_palabras = contar_palabras(frase)

    print("Cantidad de palabras:", cantidad_palabras)


def buscar_telefono(contactos):
    nombre = input("Ingrese el nombre del contacto: ")

    if nombre in contactos:
        print("Teléfono:", contactos[nombre])
    else:
        print("El contacto no existe.")


def dividir():
    try:
        n1 = float(input("Ingrese el primer número: "))
        n2 = float(input("Ingrese el segundo número: "))

        resultado = n1 / n2

        print("Resultado:", resultado)

    except ZeroDivisionError:
        print("Error: no es posible dividir entre cero.")

numeros = (8, 3, 15, 1, 10, 5)

contactos = {
    "Sofia": "4491234967",
    "Alexander": "4427650321",
    "Manuel": "4459876643"
}
opcion = 0

while opcion != 5:
    print("1. Mostrar números ordenados")
    print("2. Buscar teléfono")
    print("3. Dividir dos números")
    print("4. Analizar mensaje de texto")
    print("5. Salir")

    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:

        numeros_ordenados = sorted(numeros)

        print("Tupla original:", numeros)
        print("Números ordenados:", numeros_ordenados)

    elif opcion == 2:

        buscar_telefono(contactos)

    elif opcion == 3:

        dividir()

    elif opcion == 4:

        analizar_texto()

    elif opcion == 5:

        print("Programa finalizado.")

    else:

        print("Opción no válida. Intente nuevamente.")
```

Extra 1: Sistema de calificaciones con tuplas
Enunciado: Crea una tupla calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0). Imprime la tercera calificación, captura dos calificaciones nuevas con input() y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala de mayor a menor y muestra la suma de todos los elementos usando una función que retorne el valor.

```python
def sumar_calificaciones(calificaciones):
    return sum (calificaciones)

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0) 
print(f"La tercera calificacion es: {calificaciones [2]}")

cal1=float(input("Ingrese la primera calificacion: "))
cal2=float(input("Ingrese la segunda calificacion: "))
nuevas_calificaciones = calificaciones + (cal1,cal2)

print(f"La nueva tupla es: {nuevas_calificaciones}")

lista = list(nuevas_calificaciones)
lista.sort(reverse=True)

print(f"La lista de mayor a menor es: {lista}")

suma_final = sumar_calificaciones(nuevas_calificaciones)

print(f"La suma total de todas las calificaciones es: {suma_final}")
```

Extra 2: Agenda de contactos con búsqueda
Enunciado: Crea un diccionario agenda con tres contactos iniciales ("Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"). Captura un nuevo contacto (nombre y teléfono) con input() y agrégalo al diccionario. Luego imprime todos los nombres con keys() en una sola línea y, con una función que retorne el valor, busca y muestra el teléfono del nuevo contacto.

```python
contactos = {"Ana": "555-0101", 
             "Luis": "555-0102", 
             "Mía": "555-0103"}

nombre =input("Cual es el nombre del usuario: ")
contacto1=int(input("Ingrese el numero de contacto: "))
buscar = input("Cual es el usuario que quiere buscar: ")

contactos[nombre] = contacto1
nombres = ", ".join (contactos.keys())
print(f"El contacto fue guardado: {nombres}")

def buscar_contacto(contactos,buscar):
    return contactos.get(buscar, "No exixte")

agenda = buscar_contacto(contactos,buscar)
print(f"El telefono de {buscar} es: {agenda}")
```
